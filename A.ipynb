```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.utils import get_file
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN,LSTM,GRU,Dense
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
```

```python
DATA_URL = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"
csv_path = get_file("airline-passangers.csv",origin=DATA_URL)
df = pd.read_csv(csv_path)
series = df.iloc[:,1].values.astype("float32").reshape(-1,1)
print("Dataset Shape",series.shape)
```

```python
print(df.head())
```

```python
plt.figure(figsize=(10,4))
plt.plot(series)
plt.title("International Airline Passangers (1949-1960)")
plt.xlabel("Month")
plt.ylabel("Passangers (thousands)")
plt.tight_layout()
```

```python
scaler = MinMaxScaler(feature_range=(0,1))
scaled = scaler.fit_transform(series)
```

```python
LOOK_BACK = 12
```

```python
def create_dataset(data,look_back=1):
  X,y=[],[]
  for i in range(len(data)-look_back):
    X.append(data[i:i+look_back,0])
    y.append(data[i+look_back,0])
  return np.array(X),np.array(y)

```

```python
X,y=create_dataset(scaled,LOOK_BACK)
```

```python
split = int(len(X)*0.8)
X_train , X_test = X[:split],X[split:]
y_train , y_test = y[:split],y[split:]
print("Train shape",X_train.shape,"Test shape:",X_test.shape)
```

```python
X_train = X_train.reshape((X_train.shape[0],LOOK_BACK,1))
X_test = X_test.reshape((X_test.shape[0],LOOK_BACK,1))
print("Train shape",X_train.shape,"Test shape:",X_test.shape)
```

```python
def train_and_evaluate(model,name,epochs=100):
  model.compile(optimizer="adam",loss="mse")
  history = model.fit(X_train,y_train,validation_data=(X_test,y_test),epochs=epochs,batch_size=8,verbose=0)

  train_pred = model.predict(X_train,verbose = 0)
  test_pred = model.predict(X_test,verbose = 0)

  train_pred_inv = scaler.inverse_transform(train_pred)
  y_train_inv = scaler.inverse_transform(y_train.reshape(-1,1))
  test_pred_inv = scaler.inverse_transform(test_pred)
  y_test_inv = scaler.inverse_transform(y_test.reshape(-1,1))

  train_rmse = np.sqrt(mean_squared_error(y_train_inv,train_pred_inv))
  test_rmse = np.sqrt(mean_squared_error(y_test_inv,test_pred_inv))

  print(f"\n[{name}] Train RMSE: {train_rmse:.2f} | Test RMSE = {test_rmse:.2f}")

  plt.figure(figsize=(10,4))
  plt.plot(history.history["loss"],label="train_loss")
  plt.plot(history.history["val_loss"],label="val loss")
  plt.title(f"{name} - Loss Curve")
  plt.xlabel("Epochs")
  plt.ylabel("MSE")
  plt.legend()
  plt.tight_layout()

  return test_pred_inv , y_test_inv , test_rmse
```

```python
rnn_model =  Sequential([SimpleRNN(50,activation="tanh",input_shape=(LOOK_BACK,1)),Dense(1)])
rnn_pred,y_test_inv,rnn_rmse = train_and_evaluate(rnn_model,"Simple RNN")

lstm_model = Sequential([LSTM(50,activation="tanh",input_shape=(LOOK_BACK,1)),Dense(1)])
lstm_pred, _, lstm_rmse  = train_and_evaluate(lstm_model,"LSTM")

gru_model = Sequential([GRU(50,activation="tanh",input_shape=(LOOK_BACK,1)),Dense(1)])
gru_pred,_,gru_rmse = train_and_evaluate(gru_model,"GRU")

plt.figure(figsize = (10,5))
plt.plot(y_test_inv,label="Actual",linewidth=2)
plt.plot(rnn_pred,label=f"Simple RNN (RMSE={rnn_rmse:.1f})",linewidth=2)
plt.plot(lstm_pred,label=f"LSTM (RMSE={lstm_rmse:.1f})",linewidth=2)
plt.plot(gru_pred,label=f"GRU (RMSE={gru_rmse:.1f})",linewidth=2)
plt.title("Airline Passangers (1949-1960)")
plt.xlabel("Month")
plt.ylabel("Passangers (thousands)")
plt.legend()
```
